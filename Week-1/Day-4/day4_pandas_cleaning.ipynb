### Step 4. Create TWO derived columns

Here I havre made two derived cols, one using vectorized operation and the other using .apply().
- Derived Column 1 (vectorized)

  To calculate increase in population

````python
df["population_increase"] = df["population_2017"] - df["population_1951"]
````

````python
df[["region", "population_1951", "population_2017", "population_increase"]].head(10)
````

- Derived Column 2 (.apply())

  To classify regions

````python
def region_size(population):
    if population >= 10000:
        return "Very Large"
    elif population >= 5000:
        return "Large"
    else:
        return "Small"

df["region_size"] = df["population_2017"].apply(region_size)
````

````python
df[["region", "population_2017", "region_size"]].head(10)
````

I created two new columns.

- `population_increase` using a vectorized operation.
- `region_size` using `.apply()` to classify each region according to its 2017 population.

### Step 5. Reshape the data

````python
summary = df.groupby("region_size").agg(
    average_population=("population_2017", "mean"),
    average_growth=("growth_2017", "mean"),
    number_of_regions=("region", "count")
)
````

````python
summary
````

I grouped the dataset by `region_size` and calculated:

- average population
- average growth rate
- number of regions

This creates a summary table for each category.
