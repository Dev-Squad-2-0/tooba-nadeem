### Plot a bar chart showing average value grouped by a category column

```python
plt.bar(df["region"],df["population_2017"])
```

well I only have one category (categorical variable) which is "region" but then we have single value for each region and only once, so taking mean or average isn't meaningful. Therefore I am creating a new category to fulfill this question's requirement.

Instead of looking at the avg population by region, I am making a new column called "population_category" which classifies regions by population size, 

```python
bins = [0,5000, 10000,20000,np.inf]

labels = ["Small", "Medium", "Large", "Very Large"]

df["population_category"] = pd.cut(df["population_2017"],bins = bins, labels = labels)
```

now, we have such a category which can be utilized for calculating average.

```python
avg_population = (df.groupby("population_category")["population_2017"].mean())

# avg_population is a pandas series, and so it has index and values. when we write .index, we get Small, Medium, Large, Very Large. And when we write .vlaues, we get values.
```

now we make the barplot

```python
plt.bar(avg_population.index, avg_population.values)

plt.title("Average Population by Population Category (2017)")
plt.xlabel("Population Category")
plt.ylabel("Average Population (2017)")
```

Since each region appears only once in the dataset, calculating the average population grouped by region would not be meaningful.
To create a useful grouped bar chart, I categorized regions into four population groups: Small, Medium, Large, and Very Large.
The chart shows the average population within each category, making it easier to compare typical population sizes across different groups.

###  Plot a boxplot to visualize the distribution and spot outliers visually

```python
plt.figure(figsize=(5,5))

sns.boxplot(df["population_2017"])

plt.title("Distribution of Population in 2017")
plt.ylabel("population_2017")
```

The boxplot summarizes the distribution of the 2017 population values across all regions. It shows the median, the spread of the middle 50% of the data, and any potential outliers. Regions with exceptionally high populations appear as outliers above the upper whisker.

### Plot a correlation heatmap of all numeric columns using Seaborn

```python
# First, we select only the numeric columns.
numeric_df = df.select_dtypes(include="number")
```

```python
# Then we compute the correlation matrix.
correlation_matrix = numeric_df.corr()
```

```python
# Lastly, plot the heatmap
plt.figure(figsize=(10,8))

sns.heatmap(correlation_matrix,annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Numeric Columns")
```

The correlation heatmap shows the strength and direction of the relationships between all numeric columns. Correlation values close to 1 indicate a strong positive relationship, values close to -1 indicate a strong negative relationship, and values near 0 indicate little or no linear relationship. In this dataset, population values across different census years are expected to have strong positive correlations because regions with larger populations tend to remain relatively larger over time.
