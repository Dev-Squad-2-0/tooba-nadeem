### Urgent little cleaning of our today's dataset

```python
# Step 1: Clean the team names
# Remove leading/trailing spaces and tabs
rounds["team"] = rounds["team"].str.strip()

matches["team_name"] = matches["team_name"].str.strip()

# Standardize abbreviated team names
matches["team_name"] = matches["team_name"].replace({
    "W. Bulldogs": "Western Bulldogs"
})
```

```python
# Step 2: Also clean the opponent names (good practice)
rounds["opponent"] = rounds["opponent"].str.strip()

matches["opponent"] = matches["opponent"].str.strip()
```

```python
# Step 3: Make sure the dates have the same datatype
rounds["match_date"] = pd.to_datetime(rounds["match_date"])
matches["match_date"] = pd.to_datetime(matches["match_date"])
```

Since both datasets have a team column but with different name ( rounds has: '**team**' while matches has '**team_name**'), we have to change one to merge correctly!

```python
#Step fix team cloumn
matches.rename(
    columns={"team_name": "team"},
    inplace=True
)
```

## Quickly saving cleaned dataset

```python
rounds.to_csv("cleaned_round_by_round_stats_v2.csv", index=False)

matches.to_csv("cleaned_team_matches.csv", index=False)
```

## Finally Merging

```python
# merge
merged = pd.merge(
    rounds,
    matches[[
        "year",
        "round",
        "team",
        "match_date",
        "home_away",
        "venue",
        "crowd"
    ]],
    on=["year", "round", "team", "match_date"],
    how="left",
    indicator=True
)
```

### 4. Contextual Analysis
Using the enriched dataset, answer questions such as:
    
- Do players perform better at home or away?
- Does crowd size appear to influence fantasy points?
- Which venues produce the highest average player performance?
    
Support every conclusion with evidence.

```python
# Q1. Do players perform better at home or away?
home_away_analysis = (
    merged.groupby("home_away")["fantasy_points"]
    .mean()
    .sort_values(ascending=False)
)

home_away_analysis
```

```python
# visualization
plt.figure(figsize=(5,4))

home_away_analysis.plot(kind="bar")

plt.title("Average Fantasy Points: Home vs Away")
plt.xlabel("Match Location")
plt.ylabel("Average Fantasy Points")

plt.tight_layout()
plt.savefig("home_vs_away.png")
plt.show()
```

### Home vs Away Performance

Players scored an average of **66.49** fantasy points in **home** matches and **64.00** fantasy points in **away** matches.

This suggests that players generally performed **slightly better at home** than away. However, the difference of approximately **2.5 fantasy points** is relatively small, indicating that home advantage has only a modest impact on player fantasy performance.

```python
# Q2. Does crowd size influence fantasy points?
# correlation
merged[["crowd", "fantasy_points"]].corr()
```

```python
# scatter plot
plt.figure(figsize=(7,5))

plt.scatter(
    merged["crowd"],
    merged["fantasy_points"],
    alpha=0.3
)

plt.title("Crowd Attendance vs Fantasy Points")
plt.xlabel("Crowd Attendance")
plt.ylabel("Fantasy Points")

plt.tight_layout()
plt.savefig("crowd_vs_fantasy.png")
plt.show()
```

### Crowd Attendance Analysis

The correlation coefficient between **crowd attendance** and **fantasy points** was **0.0129**.

This indicates **almost no relationship** between crowd size and player fantasy performance.

The scatter plot also shows **no clear trend**, suggesting that crowd attendance has **little to no influence** on fantasy points.

## Data Quality Report

### Merge Keys Used

A composite key consisting of **year**, **round**, **team**, and **match_date** was used to merge the datasets. A composite key was necessary because no single column uniquely identified each team match.

### Challenges Encountered

During the integration process, inconsistencies were found in the `team_name` column of the Team Match dataset. Some team names contained leading or trailing whitespace, while others used abbreviations such as **"W. Bulldogs"** instead of **"Western Bulldogs"**.

These issues were resolved by:
- Removing extra whitespace using `.str.strip()`.
- Standardizing abbreviated team names using `.replace()`.

### Data Quality Issues Discovered

After cleaning the team names, all player records matched successfully with the Team Match dataset.

No duplicate player records were introduced during the merge, and the total number of player records remained unchanged.

The `crowd` column still contained missing values, indicating missing attendance information in the original source dataset rather than merge errors.

### Assumptions Made During Integration

The integration assumed that the combination of **year**, **round**, **team**, and **match_date** uniquely identifies each team match.

It was also assumed that standardized team names represented the same clubs across both datasets.
