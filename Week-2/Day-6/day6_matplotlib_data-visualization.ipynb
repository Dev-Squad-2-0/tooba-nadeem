## Implementation Task
Recreate 6 different chart types, from a single dataset, each with a one-line “what this tells us” caption.
- line
- bar
- histogram
- boxplot
- scatter
- heatmap 

## 1) LINE CHART

```python
plt.plot(df["age"],df["bmi"])

plt.title("Line Chart showing trend for bmi wrt age")
plt.xlabel("age")
plt.ylabel("bmi")

plt.grid(True)
```

The resulting line chart fails to represent anything meaningful since the ages do not appear in proper sequence. On the other hand if we take average BMI at each age and then plot a line chart, that would be more meaningful hopefully.

```python
avg_bmi = df.groupby("age")["bmi"].mean()
plt.plot(avg_bmi.index, avg_bmi.values)

plt.title("Line Chart showing trend of Avg BMI at each Age")
plt.xlabel("Age")
plt.ylabel("Avg_BMI")

plt.grid(True)
```

### What this tells us?
Average BMI tends to increase with age, although the trend is not perfectly consistent.

## 2) BAR CHART

```python
# avg bmi by sex
avg_bmi = df.groupby("sex")["bmi"].mean()
plt.bar(avg_bmi.index, avg_bmi.values)

plt.title("Bar Chart Showing Avg BMI by Sex")
plt.xlabel("Sex")
plt.ylabel("Avg_BMI")
```

### What this tells us?
The average BMI is similar for both males and females, with only a slight difference between them.

## 3) HISTOGRAM

```python
# histogram for charges distribution
plt.hist(df["charges"], bins=20, edgecolor="black")

plt.title("Histogram showing Charges Distribution")
plt.xlabel("Charges")
plt.ylabel("Frequency")
```

### What this tells us?
Most individuals have lower insurance charges, while a smaller number have much higher charges, resulting in a right-skewed distribution.

## 4) BOX PLOT

```python
sns.boxplot(data=df, x="smoker", y="charges")

plt.title("Boxplot showing comparison of insurance charges by Smoking Status")
```

### What this tells us?
Smokers have a much higher median insurance charge and greater variability than non-smokers.

## 5) SCATTER PLOT

```python
plt.scatter(df["age"], df["charges"])

plt.title("Scatter Plot of Age vs Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")

plt.show()
```

### What this tells us?
Insurance charges generally tend to increase with age, although there is considerable variation.

## 6) HEATMAP

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="Blues")

plt.title("Correlation Heatmap of Numeric Variables")
```

### What this tells us?
Age shows the strongest positive correlation with insurance charges among the numeric variables, while the other correlations are relatively weak.
