# Implementation Task

## Question 1
### Business Question

Which teams have played the most games?

```python
# It stores the player's entire career team history, usually as a set or list.
df["player_teams"]
```

```python
# One player's statistics for one team in one season."Which team was this player playing for in this particular record?"
df["team"]
```

```python
# games_played is associated with the player
df["games_played"]
```

```python
# Because teams don't have a games_played column. Players do.
# Q: Which teams have played the most games?
total_games_by_team = df.groupby("team")["games_played"].sum().sort_values(ascending=False)
print(total_games_by_team)
```

```python
plt.barh(total_games_by_team.index, total_games_by_team.values)
plt.gca().invert_yaxis()

plt.title("Bar Chart Showing total games by team in descending order")
plt.xlabel("Games")
plt.ylabel("Team")

# save the plot
plt.tight_layout()
plt.savefig("q1_total_games_by_team.png", dpi=300, bbox_inches="tight")
```

### Observations

- Geelong Cats have the highest total player-games (16,834), followed closely by West Coast Eagles (16,518) and Essendon Bombers (16,462).
- Most established teams have similar total player-games, generally ranging between 14,000 and 17,000.
- Brisbane Bears and Fitzroy Lions have the lowest totals, while Gold Coast Suns and Greater Western Sydney Giants also have relatively fewer player-games, likely due to their shorter participation history in the league.

### Business Insight

Teams with the highest total player-games have demonstrated greater long-term player participation across seasons. This may indicate a longer history in the competition, providing more opportunities for player appearances and contributing to richer historical performance data.

## Question 2
### Business Question

What is the distribution of player ages (last_age)?

```python
# word "distribution" tells us histogram
plt.hist(df["last_age"],edgecolor="black")

plt.title("Histogram showing distribution of player ages(last_age)")
plt.xlabel("Ages")
plt.ylabel("Number of players")

# save the plot
plt.tight_layout()
plt.savefig("q2_age_distribution.png", dpi=300, bbox_inches="tight")
```

### Observations

- Most players have a last recorded age between approximately 27 and 32 years, with the highest frequency around age 30.
- The number of players decreases toward both the younger and older age groups, indicating that very young and very old players are less common.
- Very few players have a last recorded age above 37 years, showing that longer playing careers are relatively uncommon.

### Business Insight

The majority of players finish their careers around the age of 30, suggesting that teams primarily consist of athletes in their late 20s and early 30s. This information can help teams plan succession strategies and player recruitment as experienced players approach the later stages of their careers.

## Question 3
### Business Question

Which teams have the highest number of players?

```python
highest_num_of_players = df["player_teams"].value_counts().sort_values(ascending=False)
print(highest_num_of_players)

# top 10 for bar chart
highest_number_of_players = (df["player_teams"].value_counts().head(10))
```

```python
plt.barh(highest_number_of_players.index, highest_number_of_players.values)

plt.title("Top 10 Player Team Combinations")
plt.xlabel("Number of Players")
plt.ylabel("Player Teams")

plt.gca().invert_yaxis()

# save the plot
plt.tight_layout()
plt.savefig("q3_players_by_team.png", dpi=300, bbox_inches="tight")
```

### Observations

- `{West Coast Eagles}` has the highest number of players, followed by `{Essendon Bombers}` and `{Sydney Swans}`.
- The top 10 team combinations have relatively similar player counts, with most ranging between approximately 700 and 1,000 players.
- `{Fremantle Dockers}` has the lowest player count among the top 10 team combinations shown in the chart.

### Business Insight

Teams with larger player counts indicate a stronger historical player presence in the dataset. This information can help identify clubs with extensive player records for historical analysis and performance comparisons.

## Question 4
### Business Question

How does player weight vary across different teams?

```python
sns.boxplot(data=df,x="team",y="weight")

plt.xticks(rotation=90)

plt.title("Player Weight by Team")

# save the plot
plt.tight_layout()
plt.savefig("q4_weight_by_team.png", dpi=300, bbox_inches="tight")
```

### Observations
- The median player weight is fairly similar across most teams, generally around **85–90 kg**, indicating little variation in typical player weight.
- Most teams have a similar spread of player weights, suggesting that player weight distributions are consistent across the league.
- Several teams have outliers, with a few players weighing significantly more (above 110 kg) or less (around 60–70 kg) than most players.

### Business Insight
Player weight is relatively consistent across AFL teams, indicating that teams recruit players with similar physical characteristics. The presence of outliers suggests that some teams include players with specialized roles requiring unusually high or low body weight.

## Question 5
### Business Question

Which teams have the highest average fantasy points?

```python
avg_fantasy = (df.groupby("team")["avg_fantasy_points"].mean().sort_values(ascending=False))
avg_fantasy.head(10)
```

```python
plt.bar(avg_fantasy.head(10).index,avg_fantasy.head(10).values)

plt.title("Top Teams by Average Fantasy Points")
plt.xlabel("Team")
plt.ylabel("Average Fantasy Points")

plt.xticks(rotation=90)

plt.tight_layout()
plt.savefig("q5_avg_fantasy_points.png", dpi=300, bbox_inches="tight")
```

### Observations

- Hawthorn Hawks have the highest average fantasy points among all teams.
- The average fantasy points of the top 10 teams are very close, with only a small difference between the highest and lowest values.
- Greater Western Sydney Giants and Geelong Cats are also among the top-performing teams based on average fantasy points.

### Business Insight

The small variation in average fantasy points among the top teams suggests that player performance is relatively consistent across leading teams. Teams with slightly higher averages may have stronger overall player contributions, which can support better match performance and fantasy sports analysis.

## Question 6

## Question 7
### Business Question

How many records are available for each season?

```python
season_records = (df["year"].value_counts().sort_index())
season_records
```

```python
plt.plot(season_records.index,season_records.values,marker="o")

plt.title("Number of Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")

plt.grid(True)

plt.tight_layout()
plt.savefig("q7_records_by_season.png", dpi=300, bbox_inches="tight")
```

### Observations

- The number of records increased steadily from the early seasons and grew rapidly during the 1990s.
- From around 2000 onwards, the number of records remained relatively stable, generally ranging between 800 and 900 records per season.
- There is a noticeable increase around 2012, after which the number of records remains consistently high with only minor fluctuations.

### Business Insight

The increasing number of records over time suggests improved data collection and league growth. The stable record counts in recent seasons indicate consistent data availability, making modern seasons more suitable for reliable performance analysis and year-to-year comparisons.
